```python
import requests
import pandas as pd

# 1. Fetch data from GitHub API
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"
response = requests.get(url)
data = response.json()

# 2. Convert JSON to DataFrame
df_raw = pd.DataFrame(data['items'])

# 3. Extract required values from nested columns
df_raw['owner_name'] = df_raw['owner'].apply(lambda x: x['login'] if isinstance(x, dict) else None)
df_raw['license_name'] = df_raw['license'].apply(lambda x: x['name'] if isinstance(x, dict) and x else 'No License')

# 4. Select required columns
selected_columns = [
    'id', 'name', 'full_name', 'owner_name', 'stargazers_count',
    'forks_count', 'open_issues_count', 'language', 'license_name',
    'created_at', 'updated_at'
]
df = df_raw[selected_columns].copy()

# 5. Handle missing values & duplicates
df['language'].fillna('Unknown', inplace=True)
df['license_name'].fillna('No License', inplace=True)
df.drop_duplicates(subset=['id'], inplace=True)

# 6. Rename columns & format dates
df.rename(columns={
    'stargazers_count': 'stars',
    'forks_count': 'forks',
    'open_issues_count': 'open_issues',
    'created_at': 'created_date',
    'updated_at': 'updated_date'
}, inplace=True)

df['created_date'] = pd.to_datetime(df['created_date']).dt.strftime('%Y-%m-%d')
df['updated_date'] = pd.to_datetime(df['updated_date']).dt.strftime('%Y-%m-%d')

# 7. Save cleaned dataset as CSV
df.to_csv('github_projects.csv', index=False)

print("✅ Task 1 Completed: Dataset cleaned and saved as 'github_projects.csv'")
df.head()
```

```python
import sqlite3
import matplotlib.pyplot as plt

# 1. Create SQLite Database & Import Dataset
conn = sqlite3.connect('github_data.db')
df.to_sql('Repositories', conn, if_exists='replace', index=False)

def run_query(query, title):
    print(f"\n--- {title} ---")
    result = pd.read_sql_query(query, conn)
    print(result.head(10))
    return result

# 2. Filtering & Searching Queries (AND, OR, NOT)
query_filter = """
SELECT name, language, stars, license_name
FROM Repositories
WHERE (language = 'Python' OR language = 'Jupyter Notebook')
  AND stars > 5000
  AND NOT license_name = 'No License';
"""
run_query(query_filter, "Filtering with AND, OR, NOT")

# 3. Sorting & Limiting Queries
query_sort = """
SELECT name, owner_name, stars, forks
FROM Repositories
ORDER BY stars DESC
LIMIT 10;
"""
run_query(query_sort, "Top 10 Most Popular Repositories")

# 4. Aggregate Functions (COUNT, AVG)
query_agg = """
SELECT
    COUNT(*) AS total_repositories,
    ROUND(AVG(stars), 2) AS average_stars
FROM Repositories;
"""
run_query(query_agg, "Aggregate Calculations")

# 5. Grouping Analysis (GROUP BY, HAVING)
query_group = """
SELECT language, COUNT(*) AS repo_count
FROM Repositories
GROUP BY language
HAVING COUNT(*) >= 2
ORDER BY repo_count DESC;
"""
run_query(query_group, "Grouping Analysis by Language")

# 6. Data Visualizations (Matplotlib)

# Chart 1: Top 10 Repositories by Stars
top_10 = pd.read_sql_query("SELECT name, stars FROM Repositories ORDER BY stars DESC LIMIT 10;", conn)
plt.figure(figsize=(10, 5))
plt.barh(top_10['name'], top_10['stars'], color='#2b5c8f')
plt.xlabel('Stars Count')
plt.ylabel('Repository Name')
plt.title('Top 10 Most Popular GitHub Repositories')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Chart 2: Creation Trends
df['created_year'] = pd.to_datetime(df['created_date']).dt.year
trends = df.groupby('created_year').size()

plt.figure(figsize=(8, 4))
plt.plot(trends.index, trends.values, marker='o', color='#2e7d32', linewidth=2)
plt.xlabel('Year')
plt.ylabel('Number of Repositories Created')
plt.title('Repository Creation Trends Over Time')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

conn.close()
```
